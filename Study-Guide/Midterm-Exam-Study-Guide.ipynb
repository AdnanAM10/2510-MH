```python
import pandas as pd
grade = pd.DataFrame({
    "quiz1": [100, 84, 58],
    "quiz2": [92, 77, 65],
    "quiz3": [85, 90, 70],
})

print(grade.quiz1.std()) #sample std deviation of quiz1 column
print("****************")
print(grade[['quiz1']].std())  # same expression --> same result as above.
print("***************")
print(grade.var()) #sample var for each column
```

```python
import pandas as pd

employee = pd.DataFrame({
    "Name": ["Jame", "Kate", "Omar", "Mei", "Luis"],
    "Age": [19, 24, 21, 28, 32],
    "City": ["Austin", "Riverside", "San Jose" , "Seattle", "Portland"],  # blank city as pd.NA
    "HoursWorked": [40, 45, 35, 50, 30],
})

# Single column subsetting

#all return the same result.
print(employee[["Name"]])
print(employee['Name'])
print(employee.Name)

#multiple column subseeting

print(employee[['Name', 'City']])
```

```python
# Row subsetting

#Return the second and third rows
print(employee.iloc[1:3])  #integer based

print('************')
print(employee.loc[1:2]) #label based


#first three rows and and the columns, Name and City
print("*********")
print(employee.loc[0:2, ['Name', 'City']])
print("*********")
print(employee.iloc[0:3, [0,2]])
```
