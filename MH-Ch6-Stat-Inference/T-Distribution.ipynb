```python
from scipy.stats import t  # t stands for t-distribution

import numpy as np

mu = 12          # Population mean
x_bar1 = 12      # Lower bound sample mean
x_bar2 = 14.5    # Upper bound sample mean
s = 2.5          # Sample standard deviation
n = 10           # Sample size

# Step 1: Convert to t-scores
t1 = (x_bar1 - mu) / (s / np.sqrt(n))
t2 = (x_bar2 - mu) / (s / np.sqrt(n))

print(f"t1 = {t1:.3f}")
print(f"t2 = {t2:.3f}")

# Step 2: Degrees of freedom
df = n - 1

# Step 3: Compute probabilities
prob_lower = t.cdf(t1, df=df)    # P(t ≤ 0)
prob_upper = t.cdf(t2, df=df)    # P(t ≤ 3.162)

# Step 4: Probability between t1 and t2
prob_between = prob_upper - prob_lower

print("Probability:" , prob_between)

```
