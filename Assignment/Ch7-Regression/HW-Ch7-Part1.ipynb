## Exercise 7-9 Satatic

```python
# Import the Car_Prices data file and label it myData

import pandas as pd
myData1 = pd.read_excel('/content/drive/MyDrive/Colab Notebooks/2510/2510-MH/Assignment/Ch7-Regression/Ch7_Q9_Data_File.xlsx', sheet_name = 'Car_Prices')
myData1.head(n=8)

```

```python
# Obtaining regression output

import statsmodels.formula.api as smf
Model = smf.ols('Price ~ Age + Mileage', data = myData1).fit()
print(Model.summary())
```

```python
# Obtaining prediction

# when Age = 7,  Mileage = 70000  --> must change for your homework assignment

print(Model.predict(pd.DataFrame({'Age': [7], 'Mileage': [70000]})))
```
